# Множество (Set)

HashSet - Неупорядочное множество<br>
SortedSet - Упорядочное множество

```csharp
HashSet<int> numbers = new();
numbers.Add(1);
numbers.Add(7);
numbers.Add(3);
numbers.Add(7);
numbers.Add(1);
numbers.Add(3);
numbers
```

```csharp
SortedSet<int> numbers = new();
numbers.Add(1);
numbers.Add(7);
numbers.Add(3);
numbers.Add(7);
numbers.Add(1);
numbers.Add(3);
numbers
```

## Пересечение множеств

```csharp
HashSet<int> a = new() {1, 2, 3, 4};
HashSet<int> b = new() {3, 4, 5, 6};

a.IntersectWith(b);
a
```

## Объединение множеств

```csharp
HashSet<int> a = new() {1, 2, 3, 4};
HashSet<int> b = new() {3, 4, 5, 6};

a.UnionWith(b);
a
```

## Разница

```csharp
HashSet<int> a = new() {1, 2, 3, 4};
HashSet<int> b = new() {3, 4, 5, 6};

a.ExceptWith(b);
a
```
