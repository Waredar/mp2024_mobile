# Аргументы функции

## Ссылка на переменную

```csharp
void Func(ref int x) {
    x += 100;
    Console.WriteLine(x);
}

var x = 1;

Func(ref x);
x
```

## Только чтение переменной

```csharp
void Func1(ref readonly int x) {
    //x += 100;
    var y = x + 100;
    Console.WriteLine(y);
}
```

## Иницилизирование внутри переменной

```csharp
void Func2(out int x) {
    x = 100;
}

Func2(out int y);
y
```

```csharp
void Func5(in int x) {
    //x += 100;
    Console.WriteLine(x);
}

var x = 1;
Func5(x);
x
```
