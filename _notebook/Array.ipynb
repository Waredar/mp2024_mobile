# Массив(Array)

```csharp
//                     0          1               2          3
//                    ^4         ^3              ^2         ^1
string[] cities = {"Mocsow", "Ulynovsk", "St. Petersburg", "Minsk"};
cities
```

```csharp
foreach(var city in cities)
    Console.WriteLine(city);
```

### Копирование массива

```csharp
var citiesNew = (string[])cities.Clone();
citiesNew
```

### Изменить размерность массива

```csharp
string[] citiesNew = { };
Array.Resize<string>(ref citiesNew, 10);
Array.Copy(cities, citiesNew, 2);
Console.WriteLine(String.Join(", ", citiesNew));
```

### Энумирация

```csharp
var numbers = new int[10];

for (int i = 0, n = 10; i < 10; i++, n+=2)
    numbers[i] = n;

numbers
```

```csharp
var numbers = Enumerable.Range(10, 10).ToArray();
numbers
```
