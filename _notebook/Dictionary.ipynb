# Словарь (Dictionary)

```csharp
Dictionary<int, string> d = new();
d.Add(4, "Хорошо");
d.Add(5, "Отлично");
d.Add(3, "Удовлетворительно");

d[5] = "ОТЛИЧНО";

d[2] = "Неудовлетворительно";

d.Remove(4);

d
```

```csharp
foreach(var i in d)
    Console.WriteLine($"{i.Key} | {i.Value}");
```

```csharp
foreach(var key in d.Keys)
    Console.WriteLine(key);
```
