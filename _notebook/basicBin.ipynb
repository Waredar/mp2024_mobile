# Побитовые операции

```csharp
byte n = 0b_00010_0100;
```

```csharp
byte x = 3;
byte y = 5;

Console.WriteLine($"{x:B4}");
Console.WriteLine($"{y:B4}");
Console.WriteLine();
Console.WriteLine($"{y & x:B4} - AND");
Console.WriteLine($"{y | x:B4} - OR");
Console.WriteLine($"{y ^ x:B4} - XOR");
Console.WriteLine();
Console.WriteLine($"{x << 1:B4} - x << 1");
Console.WriteLine($"{x >> 1:B4} - x << 1");

Console.WriteLine($"{~x:B4} - NOT");
```

```csharp
var x = true;
var y = false;

Console.WriteLine(x & y);
Console.WriteLine(x | y);
Console.WriteLine(x ^ y);
Console.WriteLine(!y);
```

```csharp
for (int i = 0; i < 8; i++) {
    var r = 1 << i;
    Console.WriteLine($" 1 << {i} = {r,3} | {r, 8:B8}");
}
```

```csharp
var n = 0b_1000_0000;
for (int i = 0; i < 8; i++) {
    var r = n >> i;
    Console.WriteLine($" 1 << {i} = {r,3} | {r, 8:B8}");
}
```

## Проверка бита по маске

```csharp
byte        x = 0b_0001_1011;
byte checkBit = 0b_0001_0000;

Console.WriteLine($"{x & checkBit:B8}");
Console.WriteLine((x & checkBit) == checkBit)
```

## Проверка бита по номеру

```csharp
byte      x = 0b_0001_1011;
byte NumBit = 3;
Console.WriteLine($"{1 << NumBit:B8}");
Console.WriteLine((x & (1 << NumBit)) != 0);
```
