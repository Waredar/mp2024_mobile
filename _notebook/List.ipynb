# Список(List)
| Массив | Список |
| - | - |
| Фиксированный размер | Динамический размер |
| Элементы одного типа | Элементы разных типов |
| Быстрее | Удобнее |

## Обычный список

```csharp
ArrayList arrayList = new();
arrayList.Add(1);
arrayList.Add("Hello");
arrayList.Add(3.14);
arrayList
```

## Типизированный список

```csharp
List<int> numbers = new();
numbers.Add(1);
numbers.Add(256);
numbers.Add(-1);
numbers
```

```csharp
var numbers = Enumerable.Range(0,10).ToList<int>();
numbers
```

## Инициализация Списка

```csharp
//List<string> Cities = new() {"Mocsow", "Ulynovsk", "St. Petersburg", "Minsk"};
//List<string> Cities = "Mocsow, Ulynovsk, St. Petersburg, Minsk".Split(", ").ToList<string>();
List<string> Cities = ["Mocsow", "Ulynovsk", "St. Petersburg", "Minsk"];

```

## Сортировка списка

```csharp
List<string> Cities = ["Mocsow", "Ulynovsk", "St. Petersburg", "Minsk"];
Console.WriteLine(String.Join(", ", Cities));
Cities.Sort();
Console.WriteLine(String.Join(", ", Cities));
```

## Добавление, Удаление элементов

```csharp
List<string> Cities = ["Mocsow", "Ulynovsk", "St. Petersburg", "Minsk"];
Cities.RemoveAt(1);

Cities.AddRange(new string[] {"Novgorod", "Ufa"});
Console.WriteLine(String.Join(", ", Cities));
```
