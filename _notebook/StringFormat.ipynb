# Форматирование строк

```csharp
Console.WriteLine(Convert.ToInt32("11111111", 2));

Console.WriteLine(0xFA);
```

## Выравнивание

```csharp
var name = "Lona";

Console.WriteLine($"{name}");
Console.WriteLine($"{name, 6}");
Console.WriteLine($"{name, -6}");
```

## Дозаполнение нулями

```csharp
var n = 255;

Console.WriteLine($"{n:D6}");
Console.WriteLine($"{n:X6}");
Console.WriteLine($"{n, 10:D6}");
```

## Маска

```csharp
var n = 123456789;

Console.WriteLine($"{n:#_## ###}")
```

```csharp
var phone = 978_456_56_57;
Console.WriteLine($"{phone:+7 (###) ###-##-##}");
```
