# Перечисляемый тип (enum)

```csharp
enum SingleColor {Red = 0, Green = 1, Blue = 2}

SingleColor singleColor;
singleColor = SingleColor.Green;

Console.WriteLine(singleColor);
Console.WriteLine(singleColor.ToString("D"))
```

```csharp
enum MultyColor : byte {
    Red = 1,
    Green = 2,
    Blue = 4,
    All = Red | Green | Blue,
    RedBlue = Red | Blue,
}

MultyColor multyColor;
multyColor = MultyColor.Red;
multyColor |= MultyColor.Blue;

Console.WriteLine(multyColor);

multyColor = MultyColor.All;
Console.WriteLine(multyColor);
Console.WriteLine(multyColor.ToString("D"));

multyColor ^= MultyColor.Red;
Console.WriteLine(multyColor);
```

```csharp
enum ДеньНедели : byte
{
    Пн = 1 << 0, // 0b_0000_0001
    Вт = 1 << 1, // 0b_0000_0010
    Ср = 1 << 2, // 0b_0000_0100
    Чт = 1 << 3, // 0b_0000_1000
    Пт = 1 << 4, // 0b_0001_0000
    Сб = 1 << 5, // 0b_0010_0000 
    Вс = 1 << 6, // 0b_0100_0000
    Выходные = Сб | Вс,
}
ДеньНедели деньНедели;
деньНедели = ДеньНедели.Пн;
Console.WriteLine(деньНедели);
Console.WriteLine($"{деньНедели:D}");
```
