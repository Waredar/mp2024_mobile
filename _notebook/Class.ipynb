## Обычный конструктор

```csharp
class Person {
    // поля
    public string name;
    public int age;
    // конструктор
    public Person () : this("noname") {}
    public Person(string name) : this(name, 18) {}
    public Person(string name, int age) {
        this.name = name;
        this.age = age;
    }
    // методы
    public string Speak() => "Привет";
}

var p = new Person("Влад", 32);

var p2 = new Person();
p2
```

## Свойства

```csharp
class Person {
    // поля
    private string name;
    private int age;
    // свойства
    public string Name {
        get {
            return name;
        }
        
        set {
            name = value;
        }
    }

    public int Age { get => age; set => this.age = value; }
    // конструктор
    public Person () : this("noname") {}
    public Person(string name) : this(name, 18) {}
    public Person(string name, int age) {
        this.Name = name;
        this.Age = age;
    }
    // методы
    public string Speak() => "Привет";
}

var p = new Person();

p.Name = "Юра";
```

## Первичный конструктор

```csharp
class Person (string name, int age) {
    public string Name { get; set; } = name;
    public int Age { get; set; } = age;
}

var p = new Person("123", 20);
```

```csharp
class Person (string name, int age) {
    public string Name { get; set; } = name;
    public int Age { get; set; } = age;
    public override string ToString() => $"Person: {Name}_{Age}";
}

var p = new Person("311", 31);
Console.WriteLine(p);
```

## Деконструктор (не деструктор)

```csharp
class Person (string name, int age) {
    public string Name { get; set; } = name;
    public int Age { get; set; } = age;
    public override string ToString() => $"Person: {Name}_{Age}";
    public void Deconstruct (out string name, out int age) {
        name = Name;
        age = Age;
    }
}

var p = new Person("311", 31);

(string name, int age) = p;
.//(int age1, string name1) = p;

Console.WriteLine(name);
Console.WriteLine(age);
```

## Наследование

```csharp
class Person{
    public string Name { get; set;}

}

class Emploe : Person
{
    public string Company { get; set;}
}

var e = new Emploe() {Name = "123", Company = "Yandex"};
```

## Переобпределение метода

```csharp
class Person{
    public string Name { get; set;}
    public virtual void Print() => Console.WriteLine(Name);
}

class Emploe : Person
{
    public string Company { get; set;}
    public override void Print() { 
        // base.Print(); 
        Console.WriteLine(Name + "_" + Company);
    }
}
```

## Полеморфизм + переопределение метода + скрытие метода

```csharp
class Person(string name){
    public string Name { get; set;} = name;
    public void Print() => Console.WriteLine(Name);
    public virtual void Print1() => Console.WriteLine(Name);
}

class Emploe(string name, string company) : Person(name)
{
    public string Company { get; set;} = company;
    public new void Print() { 
        // base.Print(); 
        Console.WriteLine(Name + "_" + Company);
    }
    public override void Print1() { 
        // base.Print(); 
        Console.WriteLine(Name + "_" + Company);
    }
}

Emploe e = new("vlad", "Yandex");
e.Print();
e.Print1();

Person p = new Emploe("vlad", "Yandex");
p.Print();
p.Print1();
```
