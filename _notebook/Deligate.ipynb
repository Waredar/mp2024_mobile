# Делегат

```csharp
delegate void Work();

void Hello() => Console.WriteLine("Hello");
void HowAreYou() => Console.WriteLine("How Are you?");

// однотипные делекаты
Work work;
work = Hello;

work.Invoke(); // work();
```

## Многотипные делегаты

```csharp
Work work;

work += Hello;
work += Hello;
work -= Hello;
work += HowAreYou;

if (work != null) {
    work.Invoke();
}

work?.Invoke();
```

```csharp
delegate int MathOperation(int x, int y);

MathOperation calc;

int add(int x, int y) => x + y;
int sub(int x, int y) => x - y;
int mul(int x, int y) => x * y;

calc = add;

calc(1, 2)
```

```csharp
delegate int MathOperation(int x, int y);

int add(int x, int y) => x + y;
int sub(int x, int y) => x - y;
int mul(int x, int y) => x * y;

Dictionary<string, MathOperation> calc = new() {
    ["+"] = add,
    ["*"] = sub,
    ["-"] = mul,
};


calc["+"](1, 2)
```

```csharp
delegate int MathOperation(int x, int y);

int add(int x, int y) => x + y;
int sub(int x, int y) => x - y;
int mul(int x, int y) => x * y;

enum MathOper {Add, Sub, Mul}

Dictionary<MathOper, MathOperation> calc = new() {
    [MathOper.Add] = add,
    [MathOper.Sub] = sub,
    [MathOper.Mul] = mul,
};


calc[MathOper.Add](1, 2)
```

# Событие

```csharp
class Person {
    public delegate void MyEvent(string newValue);
    private string name;
    public string Name {
        get => name;
        // set {
        //     name = value;
        //     OnTextChange?.Invoke(name);
        // }
        set => OnTextChange?.Invoke(name = value);
    }
    public MyEvent OnTextChange;
}

var p = new Person();

p.OnTextChange += s => Console.WriteLine("new Value " + s);
p.Name = "Vlad";

p.OnTextChange = s => Console.WriteLine("aaaa");
```

```csharp
class Person {
    public delegate void MyEvent(string newValue);
    private string name;
    public string Name {
        get => name;
        // set {
        //     name = value;
        //     OnTextChange?.Invoke(name);
        // }
        set => OnTextChange?.Invoke(name = value);
    }
    public event MyEvent OnTextChange;
}

var p = new Person();

p.OnTextChange += print;

void print(string s) {
    Console.WriteLine("new Value " + s);
}

p.Name = "Vlad";

//p.OnTextChange = s => Console.WriteLine("aaaa");
```
