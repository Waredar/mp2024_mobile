# Работа с  текстовыми файлами

```csharp
using System.IO;
```

```csharp
const string TEMP_DIR = "temp";

if (! Directory.Exists(TEMP_DIR))
    Directory.CreateDirectory(TEMP_DIR);
```

```csharp
Path.Combine(TEMP_DIR, "fileTXT.txt")
```

```csharp
var filename = Path.Combine(TEMP_DIR, "fileTXT.txt");

var f = new StreamWriter(filename);
try {
    f.WriteLine("Привет");
    f.WriteLine("123");
}
finally {
    f.Close();
}
```

```csharp
var filename = Path.Combine(TEMP_DIR, "fileTXT.txt");

using (var f = new StreamWriter(filename)) {
    f.WriteLine("Привет");
    f.WriteLine("123");  
}
```

```csharp
{
    using var f = new StreamWriter(filename);
    f.WriteLine("Пока");
    f.WriteLine("123");  
}
```

```csharp
using (var f = new StreamReader(filename)) {
    var st = f.ReadToEnd();
}
```

```csharp
using (var f = new StreamReader(filename)) {
    string line;
    while((line = f.ReadLine()) != null) {
        Console.WriteLine(line);
    }
}
```

```csharp
File.WriteAllText(filename, "1111\n2131\n312")
```

```csharp
File.AppendAllText(filename, "\n" + DateTime.Now.ToString());
```

```csharp
var lines = File.ReadAllLines(filename);
Console.WriteLine(String.Join("  |  ", lines));
```
