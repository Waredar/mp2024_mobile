# Функции

```csharp
int Sum1(int a, int b) {
    return a + b;
}

int Sum2(int a, int b) => a + b;
```

```csharp
int Sum3(int a, int b = 10, int c = 100) => a + b + c;

Console.WriteLine(Sum3(1));
Console.WriteLine(Sum3(1, c:200));
Console.WriteLine(Sum3(1, c:21, b:2));
```

```csharp
void PrintArray1(int[] arr) {
    Console.WriteLine(String.Join("|",arr));
}

PrintArray1([1, 2, 3, 4, 5, 6]);
```

```csharp
void PrintArray2(params int[] arr) {
    Console.WriteLine(String.Join("|",arr));
}

PrintArray2(1, 2, 3, 4, 5, 6);
```

```csharp
void PrintArray3(params object[] arr) {
    Console.WriteLine(String.Join("|",arr));
}

PrintArray3(1, "2", "hi", 3.14);
```

```csharp
int Add(int a, int b) => a + b;
string Add(string a, string b) => a + b;
string Add(string a, int b) => a + b;
string Add(int a, string b) => a + b;

int Add( params int[] num) {
    int sum = 0;
    foreach(var n in num) {
        sum += n;
    }
    return sum;
}
```

##  Метод расширения

```csharp
static void Print(this string source) {
    Console.WriteLine(source);
}
static void Print(this string source, string border) {
    Console.WriteLine(border + source + border);
}

static string MyApper(this string source) {
    return source.ToUpper() + "!!!";
}

string st = "Строка";

st.Print();
st.Print("***");
st.MyApper()
```

```csharp
static void Print(this string[] source, string separator = " ") {
    Console.WriteLine(String.Join(separator, source));
} 
string[] cities = {"Mocsow", "Ulynovsk", "St.Petersburg", "Minsk"};

cities.Print();
```
