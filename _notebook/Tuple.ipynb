# Кортеж (Tuple)

```csharp
var x = (2,4,"hello");
Console.WriteLine(x)
```

```csharp
(int min, int max) x2 = (2,4);
Console.WriteLine(x2.min);
Console.WriteLine(x2.max);
```

```csharp
// (3)
var x3 = (
    min: 2,
    max: 4
);
    
```

```csharp
//(4)
var (min, max) = (2, 4);
Console.WriteLine(min);
Console.WriteLine(max);
```
