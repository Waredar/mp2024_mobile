# Прямоугольник

```polyglot-notebook
/// <summary>
/// Нарисовать прямоугольник
/// </summary>

void PrintRectangle(
    int width, 
    int height, 
    char symbol = '*', 
    bool isFill = true,
    char symbolClear = ' ')
{
    for (int r=0; r<height; r++) {
    if (isFill || r==0 || r==height-1)
        Console.WriteLine(new string(symbol, width));
    else
        Console.WriteLine(symbol + new string(symbolClear, width-2)+symbol);
    }
}

PrintRectangle(5,5);
PrintRectangle(5,5,isFill: false)
```

```polyglot-notebook
/// <summary>
/// Нарисовать прямоугольник
/// </summary>

string[] GetRectangle(
    int width, 
    int height, 
    char symbol = '*', 
    bool isFill = true,
    char symbolClear = ' ')
{
    List<string> result = new();

    for (int r=0; r<height; r++) {
    if (isFill || r==0 || r==height-1)
        result.Add(new string(symbol, width));
    else
        result.Add(symbol + new string(symbolClear, width-2)+symbol);
    }
    
    return result.ToArray();
}

GetRectangle(10, 6, isFill:false)
```

```polyglot-notebook
var shape = GetRectangle(10, 6, isFill:false);
//(1)
foreach (var line in shape)
    Console.WriteLine(line);
```

```polyglot-notebook
//(2)
String.Join(Environment.NewLine, shape)
```

# Ромб

```polyglot-notebook
string isFillStr(int spaceWidht, int width, char spaceSymbol, char symbol, bool isFill){
    if (isFill)
        return new string(spaceSymbol, spaceWidht) + new string(symbol, width);
    else
        return new string(spaceSymbol, spaceWidht) + symbol + new string(spaceSymbol, width - 2) + symbol;
}
```

```polyglot-notebook
string[] GetDiamond(int height, 
int width = 1,
char symbol = '*',
char spaceSymbol = ' ',
bool isFill = true
){
    List<string> result = new();
    int halfHeight = height / 2;

    for (int i = 0; i < halfHeight; i++)
        result.Add(isFillStr(halfHeight - i, width + 2 * i, spaceSymbol, symbol, isFill));
    
    if (height % 2 == 1)
        result.Add(isFillStr(0, width + 2 * halfHeight, spaceSymbol, symbol, isFill));

    for (int i = halfHeight - 1; i >= 0; i--)
        result.Add(isFillStr(halfHeight - i, width + 2 * i, spaceSymbol, symbol, isFill));

    return result.ToArray();
    }
```

```polyglot-notebook
void printDiamond(int height, 
int width = 1,
char symbol = '*',
char spaceSymbol = ' ',
bool isFill = true) {
    var shape = GetDiamond(height, width, symbol, spaceSymbol, isFill);
    Console.WriteLine(String.Join(Environment.NewLine, shape));
}

printDiamond(10);
Console.WriteLine();
printDiamond(11, 2, isFill:false)
```

# Зебра

```polyglot-notebook
string[] GetZebra(
    int height,
    int width,
    int lineWidht = 1,
    char symbol = '▮',
    char spaceSymbol = ' ',
    bool isVertical = true
) {
    List<string> result = new();

    if (isVertical)
        for (int i = 0; i < height - 1; i++)
            for (int j = 0; j < lineWidht; j++)
                if (i % 2 == 0)
                    result.Add(new string(symbol, width));
                else result.Add(new string(spaceSymbol, width));
    else
        for (int i = 0; i < height - 1; i++) {
            string s = "";
            for (int j = 0; j < height; j++)
                for (int k = 0; k < lineWidht; k++)
                    if (j % 2 == 0)
                        s += symbol;
                    else s += spaceSymbol;
            result.Add(s);
        }
    return result.ToArray();
}
```

```polyglot-notebook
void printZebra(int height, 
int width,
int lineWidht = 1,
char symbol = '▮',
char spaceSymbol = ' ',
bool isVertical = true) {
    var shape = GetZebra(height, width, lineWidht, symbol, spaceSymbol, isVertical);
    Console.WriteLine(String.Join(Environment.NewLine, shape));
}

printZebra(10,10);
Console.WriteLine();
Console.WriteLine();
printZebra(5, 5, 2, isVertical:false);
```

# Спираль

```polyglot-notebook
string[] GetSpiral(
    int height, 
    int width,
    char symbol = '▮'
) {
    char[,] spiral = new char[height, width];
    int top = 0; 
    int bottom = height - 1; 
    int left = 0; 
    int right = width - 1;

    while (top <= bottom && left <= right) {
        for (int i = left; i <= right; i++)
            spiral[top, i] = symbol;
        top++;

        for (int i = top; i <= bottom; i++)
            spiral[i, right] = symbol;
        right--;

        if (top <= bottom) {
            for (int i = right; i >= left; i--)
                spiral[bottom, i] = symbol;
            bottom--;
        }

        if (left <= right) {
            for (int i = bottom; i >= top; i--)
                spiral[i, left] = symbol;
            left++;
        }
    }

    string[] result = new string[height];
    for(int i = 0; i< height; i++) {
        char[] temp = new char[width];
        for(int j = 0; j < width; j++)
            temp[i] = spiral[i, j];
        result[i] = new string(temp);
    }

    return result;
}
```

```polyglot-notebook
void printSpiral(int height, 
int width,
int lineWidht = 1,
char symbol = '▮',
char spaceSymbol = ' ',
bool isVertical = true) {
    var shape = GetSpiral(height, width);
    Console.WriteLine(String.Join(Environment.NewLine, shape));
}

printSpiral(10,10);
```
